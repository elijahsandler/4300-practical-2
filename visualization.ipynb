```python
import plotly.express as px
import pandas as pd
```

```python
df = pd.read_csv('data_collection.csv')
df.tail()
```

```python
fig = px.scatter(df, x='prompt_length', y='response_time_sec', color='database',
                 hover_data=df.columns)
fig.update_layout(xaxis_title='Prompt Length (characters)',
                  yaxis_title='Response Time (seconds)')
fig.show()
```

```python
fig = px.histogram(df, x='response_time_sec', nbins=30)
fig.update_layout(xaxis_title='Response Time (seconds)',
                  yaxis_title='Count')
fig.show()
```

```python
fig = px.violin(df, x='llm', y='response_time_sec', color='embedding',
                 hover_data=df.columns)
fig.update_layout(xaxis_title='LLM',
                  yaxis_title='Response Time (seconds)',)
fig.show()
```

```python
fig = px.scatter(df, x='response_length', y='response_time_sec', color='prompt',
                 hover_data=df.columns)
fig.update_layout(showlegend=False, 
                  xaxis_title='Response Length (characters)',
                  yaxis_title='Response Time (seconds)')
fig.show()
```
